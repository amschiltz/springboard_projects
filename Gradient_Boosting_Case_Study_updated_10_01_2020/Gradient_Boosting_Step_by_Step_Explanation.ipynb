# Gradient Boosting: Step-by-Step Explanation

This notebook walks through the gradient boosting process in a simple, visual way. I want to understand how each tree in the ensemble learns from the errors (residuals) of the previous trees, and how the final prediction is built up.

## 1. Create a New Notebook
I start by creating a new Jupyter notebook file. This gives me a blank workspace to add code and explanations.

## 2. Add Content to the Notebook
Now I add code and markdown cells to explain and demonstrate gradient boosting. Here’s a step-by-step breakdown:
1. Start with the original target values (y).
2. Train the first tree to predict y.
3. Calculate residuals (errors): actual y minus predicted y.
4. Train the next tree to predict these residuals.
5. Repeat: each new tree focuses on the errors left by the previous trees.
6. The final prediction is the sum of all tree predictions.

### Visual Diagram
```
Original y values
      |
Train Tree 1
      |
Calculate residuals (y - Tree 1 prediction)
      |
Train Tree 2 on residuals
      |
Calculate new residuals (previous residuals - Tree 2 prediction)
      |
Train Tree 3 on new residuals
      |
... (repeat)
      |
Final prediction = sum of all tree predictions
```

### Why Each Tree Is Different
Each tree in gradient boosting is trained on a different set of target values—the residuals (errors) left by the previous trees. This means:
- The first tree tries to fit the original data.
- The second tree tries to fit the errors the first tree made.
- The third tree tries to fit the errors left after the first two trees, and so on.
Because each tree is focused on correcting what the previous trees missed, the patterns it learns are different. This creates variability and helps the ensemble improve with each step.

#### Example:
Suppose my original target values are `[3, 5, 7]`.
- Tree 1 predicts `[2, 4, 6]`. Residuals: `[1, 1, 1]` (actual - predicted)
- Tree 2 is trained on `[1, 1, 1]` and predicts `[0.5, 0.5, 0.5]`. New residuals: `[0.5, 0.5, 0.5]`
- Tree 3 is trained on `[0.5, 0.5, 0.5]` and predicts `[0.2, 0.2, 0.2]`. New residuals: `[0.3, 0.3, 0.3]`
Each tree sees a different target, so it learns something new each time.
