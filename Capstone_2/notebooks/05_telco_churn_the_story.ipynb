## 1. Introduction
This analysis uses Kaggle’s Telco Customer Churn dataset, which includes information on customers who churned within the past month, along with customer demographics, account details, subscribed services, and billing information.

The goal of this analysis was to explore patterns and drivers associated with customer churn and to identify insights that could help inform retention strategies.

## 2. What proportion of customers churn?

 ![Churn Distribution](../reports/figures/eda_churn_distribution_pct.png)

 Approximately 26.6% of customers churn. A churn rate of 26.6% is relatively high for a subscription-based telecom business and suggests marked retention challenges, warranting further investigation into the factors associated with customer churn.

## 3. Does tenure differ meaningfully between churned and retained customers?

As shown in the figure below, the overall tenure for churners is almost 10 times lower (median 10 months) than the tenure for non-churners (median 38 months). 

![Boxplot comparing tenure in months for customers who churned versus those who did not.](../reports/figures/eda_tenure_by_churn_boxplot.png)

Breaking tenure into subgroups highlights the risk even more. In the figure below, each bar shows the churn rate for a specific tenure group, with the red dashed line marking the company’s overall churn rate. Labels above each bar indicate how much higher or lower that group’s churn rate is compared to average.

Customers with the shortest tenure (0–6 months) have a churn rate 43% higher than those with the longest tenure (49+ months), showing much greater risk early in the customer lifecycle.

![Binned Tenure by Churn](../reports/figures/eda_churn_rate_tenure_group.png)

## 4. How does contract type relate to churn?

Month-to-month customers churn at significantly higher rates than those on longer-term contracts. The Month-to-month churn rate is 32% higher than 1-year contracts and 40% higher than 2-year contracts. This aligns with the previous observation that customers with short tenure have much higher churn rates than customers with longer tenure.

![Contract by Churn](../reports/figures/eda_churn_rate_contract.png)

## 5. According to the model, which customer characteristics are most strongly associated with high churn risk?

The figure below shows the top 5 features predicting churn. Month-to-month contracts significantly increase churn risk, while longer contracts, greater tenure, and having phone service reduce it. Bars to the right of zero (positive coefficients) indicate higher churn risk; bars to the left (negative coefficients) indicate lower risk. The farther a bar is from zero, the stronger its impact on churn.

## 6. Hypothesis
Customers on month-to-month contracts with shorter tenure and higher monthly charges appear to be at the highest risk of churn.

## 7. Summary
 
- 26.6% of telco customers churn within a month, indicating major retention challenges.
- Highest churn risk is seen among:
    - Customers with month-to-month contracts
    - Customers with short tenure
    - Customers with higher monthly charges
- Lower churn risk is associated with:
    - Longer contracts
    - Greater tenure
    - Having phone service

Recommendation: Focus retention strategies on new and month-to-month customers to reduce churn and improve business performance.
