## 1. Introduction
This analysis uses Kaggle’s Telco Customer Churn dataset, which includes information on customers who churned within the past month, along with customer demographics, account details, subscribed services, and billing information.

The goal of this analysis was to explore patterns and drivers associated with customer churn and to identify insights that could help inform retention strategies.

## 2. What proportion of customers churn?

 ![Churn Distribution](../reports/figures/eda_churn_distribution_pct.png)

 Approximately 26.6% of customers churn. A churn rate of 26.6% is relatively high for a subscription-based telecom business and suggests marked retention challenges, warranting further investigation into the factors associated with customer churn.

## 3. Does tenure differ meaningfully between churned and retained customers?

As shown in the figure below, the overall tenure for churners is almost 10 times lower (median 10 months) than the tenure for non-churners (median 38 months). 

![Boxplot comparing tenure in months for customers who churned versus those who did not.](../reports/figures/eda_tenure_by_churn_boxplot.png)

This becomes even more striking if you break tenure into subgroups. In the following figure, each bar shows the churn rate for a specific group. The red dashed line is the company’s overall churn rate. Numbers above each bar show the group churn rate and how much higher or lower that group’s churn rate is compared to the overall average.

The analysis shows that the customers with the shortest tenure (0-6 months) have a 43% higher churn rate than those with the longest tenure (>= 49 months), indicating much higher risk early in the customer lifecycle.

![Binned Tenure by Churn](../reports/figures/eda_churn_rate_tenure_group.png)

## 4. How does contract type relate to churn?

Month-to-month customers churn at significantly higher rates than those on longer-term contracts. The Month-to-month churn rate is 32% higher than 1-year contracts and 40% higher than 2-year contracts. This aligns with the previous observation that customers with short tenure have much higher churn rates than customers with longer tenure.
